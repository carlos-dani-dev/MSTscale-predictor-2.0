```python
import pandas as pd
```

```python
csv_path = "predictions_ccv2_mac_stw_facenetft_skt_full_grouped_b128_fold1.csv"
df = pd.read_csv(csv_path)

# As duas colunas usam valores como 'scale 5'. Extrair o número torna o
# cálculo robusto a valores ausentes e permite medir a distância ordinal.
y_true = pd.to_numeric(df["monk_scale"].astype("string").str.extract(r"(\d+)")[0])
y_pred = pd.to_numeric(df["predicted_monk_scale"].astype("string").str.extract(r"(\d+)")[0])
valid = y_true.notna() & y_pred.notna()
y_true = y_true[valid].astype(int)
y_pred = y_pred[valid].astype(int)

distance = (y_true - y_pred).abs()
metrics_ccv2 = pd.Series({
    "n_avaliaveis": len(y_true),
    "acc": (y_true == y_pred).mean(),
    "ooacc": (distance <= 1).mean(),
    "mse": ((y_true - y_pred) ** 2).mean(),
})

print(f"CCv2 — {csv_path}")
print(metrics_ccv2.to_string(float_format=lambda value: f"{value:.4f}"))
```
